```python
 #Create a working copy so the original dataset stays unchanged
air_df = df.copy()

# Convert date and time
air_df["Date Time"] = pd.to_datetime(
    air_df["Date Time"],
    errors="coerce",
    utc=True
)

# Convert pollutant columns to numeric
pollutants = [
    "NOx", "NO2", "NO", "PM10",
    "NVPM10", "VPM10", "NVPM2.5",
    "PM2.5", "VPM2.5", "O3"
]

for col in pollutants:
    air_df[col] = pd.to_numeric(
        air_df[col],
        errors="coerce"
    )

# Check date range
print("Earliest date:", air_df["Date Time"].min())
print("Latest date:", air_df["Date Time"].max())

# Check NO2 measurements
print("\nAvailable NO2 readings:", air_df["NO2"].notna().sum())
print("Missing NO2 readings:", air_df["NO2"].isna().sum())
print("Negative NO2 readings:", (air_df["NO2"] < 0).sum())
```

```python
# Select NO2 measurements
no2_data = air_df["NO2"].dropna()

# Calculate statistics
mean_no2 = no2_data.mean()
median_no2 = no2_data.median()

# Create the histogram
plt.figure(figsize=(9, 5))

plt.hist(
    no2_data,
    bins=40,
    edgecolor="black",
    alpha=0.75
)

# Mean and median lines
plt.axvline(
    mean_no2,
    linestyle="--",
    linewidth=2,
    label=f"Mean = {mean_no2:.2f}"
)

plt.axvline(
    median_no2,
    linestyle="-.",
    linewidth=2,
    label=f"Median = {median_no2:.2f}"
)

# Titles and labels
plt.title("Fig(1) Distribution of NO2 Concentrations in Bristol")
plt.xlabel("NO2 concentration (µg/m³)")
plt.ylabel("Frequency")

plt.legend()
plt.tight_layout()

# Save updated graph
plt.savefig("graph1_NO2_histogram.png", dpi=300)

plt.show()
```

```python
# Count available NO2 measurements by monitoring location

location_counts = (
    air_df.dropna(subset=["NO2"])
    .groupby("Location")["NO2"]
    .count()
    .sort_values(ascending=False)
)

display(location_counts.head(10))
```

```python
# Select one monitoring location
station = "Parson Street School"

# Filter the data
station_data = air_df[
    (air_df["Location"] == station) &
    (air_df["NO2"].notna()) &
    (air_df["Date Time"].notna())
].copy()

# Sort measurements by date
station_data = station_data.sort_values("Date Time")

# Calculate daily average NO2
daily_no2 = (
    station_data
    .set_index("Date Time")["NO2"]
    .resample("D")
    .mean()
)

print("Monitoring station:", station)
print("Number of measurements:", len(station_data))
print("First measurement:", station_data["Date Time"].min())
print("Last measurement:", station_data["Date Time"].max())
print("Number of days:", daily_no2.notna().sum())

display(daily_no2.head())
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    daily_no2.index,
    daily_no2.values,
    linewidth=1,
    label="Daily average NO2"
)

plt.title("Fig(2) Daily Average NO2 Concentrations at Parson Street School")
plt.xlabel("Date")
plt.ylabel("Average NO2 concentration (µg/m³)")

plt.legend()
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.tight_layout()

# Save graph
plt.savefig("graph2_NO2_time_trend.png", dpi=300)

plt.show()
```

```python
# Select pollutant columns
heatmap_columns = [
    "NOx", "NO2", "NO",
    "PM10", "PM2.5", "O3"
]

# Create a separate dataframe
pollutant_data = air_df[heatmap_columns].copy()

# Check available observations
print("Available observations:")
print(pollutant_data.count())

# Calculate pairwise observation counts
print("\nPairwise observation counts:")
display(pollutant_data.notna().astype(int).T.dot(
    pollutant_data.notna().astype(int)
))
```

```python
# Calculate Pearson correlation
correlation_matrix = pollutant_data.corr(
    method="pearson",
    min_periods=100
)

display(correlation_matrix.round(2))
```

```python
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Fig(3) Correlation Heatmap of Air Pollutants in Bristol")
plt.tight_layout()

# Save graph
plt.savefig("graph3_pollutant_correlation_heatmap.png", dpi=300)

plt.show()
```

```python
# Extract unique pollutant pairs
correlation_pairs = (
    correlation_matrix
    .where(np.triu(np.ones(correlation_matrix.shape), k=1).astype(bool))
    .stack()
    .sort_values(ascending=False)
)

print("Pollutant correlation pairs:")
display(correlation_pairs)
```
